```python
class Human:
    # 初期化メソッドで直接名前と年齢を受け取り、属性に代入する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 情報を出力するための専用メソッド
    def check_adult(self):
      if self.age >= 20:
        print(self.name + "は大人です")
      else:
        print(self.name + "は大人ではない")

tanaka = Human("田中", 19)
yamada = Human("山田",23)
suzuki = Human("鈴木",34)

list1 = [tanaka , yamada , suzuki]

# for i in leng(list1):
#   check_adult(list1)

# リストの要素（Humanオブジェクト）を1つずつ取り出して処理します
for human in list1:
  human.check_adult()
```
