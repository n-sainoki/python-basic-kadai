```python
class Human:
    # 初期化メソッドで直接名前と年齢を受け取り、属性に代入する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 情報を出力するための専用メソッド
    def print_info(self):
        print(self.name, self.age)

# インスタンス化のタイミングで属性の値を設定する
tanaka = Human("tanaka", 19)

# 属性にアクセスして出力する
tanaka.print_info()
```
